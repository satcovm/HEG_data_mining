# Quantitative analysis
HR Datas quantitative analysis

```r
myData <- read.table (file="HR_prediction-all.csv",sep=",", header=T)
```

```r
# to be transformed into parameters
indexTargetVariable <- 8 
qualitativeColumns <- list(7,9,10,11)

#func body
numberColumns <- dim(myData)[2]
targetName <- names(myData)[indexTargetVariable]

for(index in qualitativeColumns ){
    if (index == indexTargetVariable) next

    t(prop.table(table(myData[,c(index,indexTargetVariable)])))
    barplot(t(prop.table(table(myData[,c(index,indexTargetVariable)]))), legend=T, main=paste("Joint probability: P(", targetName , "," , names(myData)[index], ")"))
}
```
